## 🔵 BÖLÜM 1: IBM Qiskit ile Evrensel Kapılara Giriş

Klasik bilgisayarlarda bitler sürekli $0$ ve $1$ durumundadır. Kuantum'da ise sonsuz uzaya yayılabilen Qubit'ler kullanırız. Okların (vektörlerin) yönünün dönmesini sağlayan temel matematiksel makinelere **Kuantum Kapıları** (Quantum Gates) denir.

* **X Kapısı (NOT):** Qubit'i $180^{\circ}$ ters çevirir ($|0\rangle \rightarrow |1\rangle$).
* **Y Kapısı:** İmajiner (kompleks sayı barındıran) bir dönüş eklentisi yapar.
* **Z Kapısı:** İhtimali değiştirmez ancak fazı çevirir.
* **T Kapısı:** Z ekseni etrafında sadece $\pi / 4$ ($45^{\circ}$) kadarlık (çok cüzi) bir hassas dönüş sağlar.

Aşağıdaki koda tıklayın ve **Shift + Enter** basarak IBM modelinde devremizi çizin:

```python
from qiskit import QuantumCircuit

# Toplam 4 Qubit içeren bir kuantum devresi (anakart) başlat.
qc = QuantumCircuit(4)

# Devreye sırayla kapılarımızı yerleştiriyoruz:
qc.x(0)   # 0. tele X kapısı
qc.y(1)   # 1. tele Y kapısı
qc.z(2)   # 2. tele Z kapısı 
qc.t(3)   # 3. tele T kapısı

# Ekran üzerinde devrenin ASCII şeması çizdiriliyor:
print("IBM Qiskit Kuantum Devre Şeması:\n")
print(qc.draw(output='text'))
```

## 🟠 BÖLÜM 2: Google Cirq ile Kuantum Zarı Atma (Gerçek Rastgelelik)

Elinizdeki akıllı telefonlar tamamen *gerçek* bir rastgele sayı üretemez (hep bir matematik formülü üzerinden sizi kandırır). Ancak **Kuantum Süperpozisyonu** doğası gereği felsefi olarak mutlak öngörülemezdir!

**Hadamard (H) Kapısı**, bir $|0\rangle$ okunu o saniyeliğine iki paralel evrene bölerek aynı anda hem Yazı (0) hem Tura (1) olduğu ($|\psi\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$) o muazzam hayalet konuma çeker.

Parçacık ölçüldüğünde dalga çöker ve gerçeklik seçilir. Aşağıda kuantum zarını tam 20 kez atıyoruz:

```python
import cirq

# 1 adet Qubit (Kuantum Parçacığı) oluştur
q = cirq.NamedQubit("Kuantum_Zari")
circuit = cirq.Circuit()

# 1- Qubiti %50 Tura (1) | %50 Yazı (0) olacak şekilde havaya fırlat (Süperpozisyon!)
circuit.append(cirq.H(q))

# 2- Yere düştüğünde kutuyu açıp sonucu ölç!
circuit.append(cirq.measure(q, key='sonuc'))

print("Kusursuz Zar Atma Devremiz:")
print(circuit)

# 20 kez deneme yapan Sanal Fizik Laboratuvarı Engine'ini aç
simulator = cirq.Simulator()
res = simulator.run(circuit, repetitions=20)

print("\nZari 20 kez pes pese havaya firlattik (Ölçümler):")
print(res.measurements['sonuc'].flatten())
```

## 🟠 BÖLÜM 3: Uzaktan Hayalet Etki (Kuantum Dolanıklık & CNOT Kapısı)

Einstein'ın varlığını kabullenmekte ezelden beri çok zorlandığı olay! Eğer iki ayrı qubit'i **CNOT (Kontrollü-NOT)** kapısı ile birleştirirsek, bu parçacıklar uzay-zaman dokusunu aşarak birbiriyle saniyesinde "telepatik" bir bağ kurar. Kuantum matematiğinde bu duruma *Bell Durumu (Bell State)* adı verilir: $|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$

Kutu açıldığında Qubitlerden biri "Tura" geldiyse, diğer qubit bir diğer galakside (Dünya'dan milyonlarca ışık yılı uzakta) olsa bile *anında* "Tura" gelir! Çıkan ikili rakamlar rastgele ama hizalıdır; yani daima ya `0 0` ya da `1 1` olmak *zorundadır*.

```python
import cirq

# Yan Yana 2 Qubit Oluştur (Dizi 0 ve Dizi 1)
q0, q1 = cirq.LineQubit.range(2)
circuit = cirq.Circuit()

# 1. Aşama: Birinci qubiti yarı ölü yarı diri hayalet sürece sok (H Kapısı)
circuit.append(cirq.H(q0))

# 2. Aşama: CNOT ile q0 ve q1'i evlilik bağıyla birleştir (Dolanıklığı Başlat!)
circuit.append(cirq.CNOT(q0, q1))

# İki parçacığı da ölç, değerini sonradan öğrenmek üzere kaydet
circuit.append(cirq.measure(q0, q1, key='deney'))

print("Dolanıklık (Bell Durumu) Semasi:")
print(circuit)

simulator = cirq.Simulator()
sonuc = simulator.run(circuit, repetitions=15) # 15 defa dene

print("\n15 Ayri Deneyin Ciktisi (Lutfen alt alta 0 ve 1'lerin HIC BOZULMADAN nasil eslestigine dikkat edin):\n")
print(sonuc)
```

## 🟠 BÖLÜM 4: Işınlanma Gerçek Oluyor! (Quantum Teleportation Protokolü)

Fizikteki bilim kurguyu kodlarla gerçeğe dönüştürüyoruz! Fiziksel bir maddenin *kendisini* değil ama **İçerisindeki ham bilgi formasyonunu (kuantum durumunu)** mesafeler arası yollayıp diğer taraftaki rastgele bir maddedeki yapıyı değiştirerek o parçacığın yerel kopyasını üretiyoruz!

**Mekanizma Olay Örgüsü:** 
1. Işınlanacak olan şifreli **Gizli Mesaj** kuantum parçacığını bir istasyon (Alice) alır.
2. Dolanık bağı sayesinde Alice bu ölçümü yaptığı anda kendi elindeki mesaj parçalanıp fiziksel yok olur.
3. Ancak bu dalga formasyonu ışık yılı ötedeki Bob istasyonuna fırlatılmış olur. Bob bu hasarlı/şifreli formasyonu kendi algoritmalarıyla düzeltir ve Gizli Mesaj'ın orijinal 3D verisine ulaşmış olur!

```python
import cirq

# 3 Aktör Teli Lazım: Mesaj (Alice'in elinde), Alice'in Agı, Bob'un Agı
msg, alice, bob = cirq.LineQubit.range(3)
circuit = cirq.Circuit()

print("---- Işınlanma Deneyi Basiyor! ----\n")

# ADIM 1: Yollanacak bilgiyi manipüle et (Mesajı [1] yaptık. İsterseniz YZ vurabilirsiniz)
circuit.append(cirq.X(msg))

# ADIM 2: Dolanıklık Köprüsü (Bob ve Alice arasında ağ oluştur)
circuit.append(cirq.H(alice))
circuit.append(cirq.CNOT(alice, bob))

# ADIM 3: Işınlanma Süreci! Alice elindeki mesaj paketini ağ tüneline yollayıp MÜHÜRLÜYOR (Olcum = imha)
circuit.append(cirq.CNOT(msg, alice))
circuit.append(cirq.H(msg))
circuit.append(cirq.measure(msg, alice))

# ADIM 4: Parçalanan radyoaktif bilgi ağı geçip Bob'un kutusuna indi!
# Bob, Alice'in ölçtüğü (imha olan) mesajın son formasyonuna göre parçasını yeniden derler:
circuit.append(cirq.CNOT(alice, bob))
circuit.append(cirq.CZ(msg, bob))

# Bob kutuyu kendi laboratuvarında açtı...
circuit.append(cirq.measure(bob, key='isınlanan_sonuc'))

print("TAM IŞINLANMA EKSİKSİZ PROTOKOL DEVRESİ:\n")
print(circuit)

# Ve beklenen an... Gerçekleşti mi?
simulator = cirq.Simulator()
result = simulator.run(circuit, repetitions=1)
print(f"\nIşınlama Bitti! Bob uzaktaki istasyonda kutuyu acti ve Gelen Mesaji Gordu: [{result.measurements['isınlanan_sonuc'][0][0]}]")
print("Islem Mukemmel Basarili! Mesaj [1] isinlanmistir!")
print("Bilim harika degil mi?")
```
